```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model = YOLO('yolov8n-seg.pt')
```

```python
results_1 = model('photo_1.jpg')
results_2 = model('photo_2.jpg')
```

```python
annotated_image_1 = results_1[0].plot()
annotated_image_1 = cv2.cvtColor(annotated_image_1, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(cv2.cvtColor(results_1[0].orig_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.imshow(annotated_image_1)
plt.axis('off')
plt.show()
```

```python
annotated_image_2 = results_2[0].plot()
annotated_image_2 = cv2.cvtColor(annotated_image_2, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(cv2.cvtColor(results_2[0].orig_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.imshow(annotated_image_2)
plt.axis('off')
plt.show()
```
