```python
from keras.datasets import imdb
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense, Bidirectional, Dropout, Input
from sklearn.metrics import classification_report
from keras.callbacks import EarlyStopping
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
max_length = 400
X_train_pad = pad_sequences(X_train, maxlen=max_length)
X_test_pad = pad_sequences(X_test, maxlen=max_length)
```

```python
callback = EarlyStopping(monitor='val_loss', patience=2)
```

```python
model = Sequential([
    Input((400,)),
    Embedding(input_dim=10000, output_dim=64),
    Bidirectional(LSTM(64, dropout=0.2)),
    Dropout(0.4),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x=X_train_pad, y=y_train, epochs=15, validation_split=0.2, callbacks=[callback])
```

```python
y_pred_probs = model.predict(X_test_pad)
y_predict = (y_pred_probs > 0.5).astype(int)
```

```python
model.evaluate(x=X_test_pad,y=y_test)
```

```python
print(classification_report(y_test, y_predict))
```
