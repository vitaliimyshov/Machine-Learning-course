```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('adult.data.csv')
data.head()
```

**1. Скільки чоловіків и жінок (ознака *sex*) представлено в цьому наборі даних?**

```python
data['sex'].value_counts(dropna=False)
```

Відповідь: 21790 чоловіків, 10771 жінок

**2. Який середній вік (ознака *age*) жінок?**

```python
age_mean = data[data['sex'] == 'Female']['age'].mean()
print(age_mean.round(2))
```

**3. Яка доля громадян Німеччини (ознака *native-country*)?**

```python
(data['native-country'] == 'Germany').mean()
```

Відповідь: ≈0.004

**4-5. Які середні значення та средньоквадратичні відхилення віку тих, хто отримує більше 50K на рік (ознака *salary*) і тих, хто отримує менше 50K на рік?**

```python
age_avg_1 = data[data['salary'] == '>50K']['age'].mean()
age_std_1 = data[data['salary'] == '>50K']['age'].std()
age_avg_2 = data[data['salary'] == '<=50K']['age'].mean()
age_std_2 = data[data['salary'] == '<=50K']['age'].std()
print(age_avg_1)
print(age_std_1)
print(age_avg_2)
print(age_std_2)
```

```python
Відповідь: вік тих, хто заробляє більше 50К: 44.25 ± 10.52
           вік тих, хто заробляє менше 50К: 36.78 ± 14.02
```

**6. Чи правда, що люди, які отримують більше 50k, мають як мінімум вищу освіту? (ознака *education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters* или *Doctorate*)**

```python
data_1 = data[['salary', 'education']][data['salary'] == '>50K']
data_1['education'].isin(['Bachelors', 'Prof-school', 'Assoc-acdm', 'Assoc-voc', 'Masters', 'Doctorate']).mean()
```

```python
Відповідь: оскільки лише приблизно 58 відсотків людей, що заробляють більше 50К, мають вищу освіту, то це твердження є хибним.
```

**8. Серед кого більша доля багатіших (>50K): серед одружених чи не одружених чоловіків (ознака *marital-status*)? За одружених вважаємо тих, у кого *marital-status* починається із *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), всі інші - не одружені.**

```python
data_marriage = data[['salary', 'marital-status']][(data['sex'] == 'Male')]
data_married = data_marriage[data_marriage['marital-status'].isin(['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse'])].value_counts('salary', normalize=True)
data_else = data_marriage[~ data_marriage['marital-status'].isin(['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse'])].value_counts('salary', normalize=True)
print(data_married)
print(data_else)
```

```python
Відповідь: доля багатіших людей серед одружених чоловіків
```

**9. Яку максимальну кількість годин в тиждень працює людина (ознака *hours-per-week*)? Скільки людей працюють таку кількість годин і який відсоток серед них багатих?**

```python
max_hours = data['hours-per-week'].max()
print('Кількість годин:', max_hours)
people_data = data[data['hours-per-week'] == max_hours]['salary']
print('Кількість людей: ', len(people_data))
data[data['hours-per-week'] == max_hours]['salary'].value_counts(normalize=True)
```

```python
Відповідь: так працюють 85 людей, приблизно 29 відсотків з них багаті.
```
