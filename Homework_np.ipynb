```python
import random 
import numpy as np
# Знайти в датасеті таргет. Завантажити датасет без таргета
data = np.genfromtxt('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', dtype=float, delimiter=',', usecols=(0, 1, 2, 3))
print(data)
```

```python
# Перетворити колонки, що залишились в 2D масив (або впевнитись, що це уже 2D масив)
print(data.ndim)
print(data.shape)
```

```python
# Порахувати mean, median, standard deviation для 1-ї колонки
col_1 = data[:, 0]
print("Mean: ", np.mean(col_1))
print("Median: ", np.median(col_1))
print("Standard deviation: ", np.std(col_1))
```

```python
"""Вставити 20 значень np.nan на випадкові позиції в масиві 
(при використанні звичайного рандому можуть накластись позиції, тому знайти рішення, яке гарантує 20 унікальних позицій)"""
# data = data.flatten()
# random.seed(42)
# rand_num = random.sample(range(len(data)), 20)
# data[rand_num] = np.nan # Перше, що спало на думку
# data = data.reshape(150, 4)
# data
np.random.seed(17)
indices = np.random.choice(len(data.flat), size=20, replace=False)
data.flat[indices] = np.nan
data
```

```python
# Знайти позиції вставлених значень np.nan в 1-й колонці
col_1_new = data[:, 0]
print(col_1_new)
np.where(np.isnan(col_1_new))
```

```python
# Замінити всі значення np.nan на 0
data[np.where(np.isnan(data))] = 0
data
```

```python
# Відфільтрувати массив за умовою: значення в 3-й колонці > 1.5 та значения в 1-й колонці < 5.0 (зберегти у іншу змінну)
data_filtered = data[(data[:, 0] < 5) & (data[:, 2] > 1.5)]
data_filtered
```

```python
# Розбити масив по вертикалі на 2 рівні частини (не використовувати абсолютні числа, мають бути два массиви по 4 колонки)
data_1, data_2 = np.split(data, 2)
print(data_1)
print(data_2)
```

```python
# Відсортувати обидва массиви по 1-й колонці: 1-й за збільшенням, 2-й за зменшенням
data_1[:, 0].sort()
data_2[:, 0].sort(descending=True)
print(data_1)
print(data_2)
```

```python
# Зібрати обидва массиви в одне ціле
data_combined = np.concatenate([data_1, data_2], axis=0)
data_combined
```

```python
# Порахувати всі унікальні значення в массиві та вивести їх разом із кількістю
uniqs, count = np.unique(data_combined, return_counts=True)
print(np.dstack((uniqs, count)))
```

```python
# Знайти найбільш часто повторюване значення в массиві
mode = uniqs[np.argmax(count)] 
print(mode) 
```

```python
# Написати функцію, яка б множила всі значення в колонці, які менше середнього значения в цій колонці, на 2, і ділила інші значення на 4.
def func(array):
    mean = np.mean(array)
    return np.where(array < mean, 2*array, array/4)
```

```python
# Застосувати отриману функцію до 3-ї колонки. Функція має прийняти лише одну колонку і повернути колонку із зміненими значеннями
data_combined[:, 2] = func(data_combined[:, 2])
data_combined
```
